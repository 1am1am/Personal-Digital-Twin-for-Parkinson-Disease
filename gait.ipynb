```python
import cv2
import numpy as np

TILE_CM = 80

# 4 góc ô gạch (pixel gốc 1080x1920)
tile_pixels = np.float32([
    [168, 1159],  # P1: trên-trái
    [542, 1190],  # P2: trên-phải
    [455, 1374],  # P3: dưới-phải
    [1, 1331],  # P4: dưới-trái
])

# Tọa độ thực tế (cm)
tile_real = np.float32([
    [0,       0      ],
    [TILE_CM, 0      ],
    [TILE_CM, TILE_CM],
    [0,       TILE_CM],
])

# Tính Homography
H, mask = cv2.findHomography(tile_pixels, tile_real)

def pixel_to_cm(px, py):
    pt = np.array([[[float(px), float(py)]]], dtype=np.float32)
    result = cv2.perspectiveTransform(pt, H)
    return float(result[0][0][0]), float(result[0][0][1])

# Ví dụ dùng:
# x_cm, y_cm = pixel_to_cm(foot_x, foot_y)
```
